```python
import numpy as np
import matplotlib.pyplot as plt

import scipy.signal as signal

#período com frequência de amostragem
t = 1/44100

#frequência_cosseno = 500
#frequência_amostragem = 44.1 kHz
#duração = 1.0

arr = np.arange (0, 1, t, ftype = None)
x = np.cos(2 * np.pi * 500 * arr)

frequencias, espectro = signal.periodogram(x, 44100)

#plotar o espectro do cosseno
plt.figure(figsize = (10,4))
plt.plot(frequencias, espectro)
plt.xlim(0,150)
plt.grid(True)
plt.show()
```
