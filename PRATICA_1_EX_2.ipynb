```python
import numpy as NP
from scipy.signal import fftconvolve

def calcular_resposta_impulso(sinal_original, sinal_gravado):
  n_fft = 2 ** int(NP.ceil(NP.log2(len(sinal_original)+len(sinal_gravado) - 1)))

  X = NP.fft.fft(sinal_original, n_fft)
  Y = NP.fft.fft(sinal_gravado, n_fft)

  eps = 1e-10
  H = Y / (X +eps)

  h = NP.real(NP.fft.ifft(H))

  h = h[:len(sinal_gravado) - len(sinal_original) + 1]
  return h / NP.max(NP.abs(h))
```
