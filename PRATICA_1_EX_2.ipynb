```python
import numpy as NP
import matplotlib.pyplot as PLT

x = NP.arange (0, 5, 1/44100, dtype=float)
def chirp(tempo) :
  y = NP.cos (2 * NP.pi * [500 * tempo + (10000 + 500)/(2*10000) * tempo ** 2])
  return y

t = 5
newY = chirp(t)

PLT.plot(x, newY)
PLT.show()
```
