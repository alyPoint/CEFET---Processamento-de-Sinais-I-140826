## **Exercício 2**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import chirp

fs = 44100                                                           # Frequência de amostragem

# nome do intervalo = np.arange(inicio, fim, passo)
t = np.arange(0, 5, 1/fs)                                            # Intervalo de tempo e amostragem

# nome do sinal = chirp(intervalo de tempo, frequencia inicial, tempo final, frequencia final, método)
sinal = chirp(t, 500, 5, 10000, method = 'linear')

#Primeiro gráfico
#plt.subplot(numero de linhas, numero de colunas, qual linha estamos) -> para criar vários gráficos em linhas e/ ou colunas diferentes
plt.subplot(3, 1, 1)
plt.plot(t, sinal)

#Segundo gráfico
plt.subplot(3, 1, 2)
plt.plot(t, sinal)
plt.xlim(0, 0.02)

#Terceiro gráfico
plt.subplot(3, 1, 3)
plt.plot(t, sinal)
plt.xlim(5 - 0.002, 5)

plt.show()
```
