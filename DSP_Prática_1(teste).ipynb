```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import chirp

fs = 44100                                                           # Frequência de amostragem

# nome do intervalo = np.arange(inicio, fim, passo)
t = np.arange(0, 5, 1/fs)                                            # Intervalo de tempo e amostragem

# nome do sinal = chirp(intervalo de tempo, frequencia inicial, tempo final, frequencia final, método)
sinal = chirp(t, 500, 5, 10000, method = 'linear')

#Primeiro gráfico
#plt.subplot(numero de linhas, numero de colunas, qual linha estamos) -> para criar vários gráficos em linhas e/ ou colunas diferentes
plt.subplot(3, 1, 1)
plt.plot(t, sinal)

#Segundo gráfico
plt.subplot(3, 1, 2)
plt.plot(t, sinal)
plt.xlim(0, 0.02)

#Terceiro gráfico
plt.subplot(3, 1, 3)
plt.plot(t, sinal)
plt.xlim(5 - 0.002, 5)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Configuração dos parâmetros do chirp
f0 = 500          # Frequência inicial (Hz)
f1 = 10000        # Frequência final (Hz)
T = 5.0           # Duração (segundos)
fs = 44100        # Frequência de amostragem (Hz)

# 2. Criação do vetor de tempo usando np.linspace
num_amostras = int(fs * T)
t = np.linspace(0, T, num_amostras, endpoint=False)

# 3. Cálculo da fase e do sinal do chirp linear
# Fórmula: phi(t) = 2 * pi * (f0 * t + ((f1 - f0) / (2 * T)) * t^2)
fase = 2 * np.pi * (f0 * t + 0.5 * (f1 - f0) * (t**2) / T)
sinal_chirp = np.sin(fase)

# 4. Construção dos gráficos
plt.figure(figsize=(12, 8))

# Gráfico 1: Visão geral dos 5 segundos
plt.subplot(3, 1, 1)
plt.plot(t, sinal_chirp, color='blue')
plt.title('Chirp no Domínio do Tempo (Sinal Completo - 5s)')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.grid(True)

# Gráfico 2: Zoom no Início (Frequência Baixa - 500 Hz)
plt.subplot(3, 1, 2)
plt.plot(t, sinal_chirp, color='darkorange')
plt.xlim(0, 0.02)  # Mostra apenas os primeiros 20 milissegundos
plt.title('Zoom no Início: Frequência Baixa (~500 Hz)')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.grid(True)

# Gráfico 3: Zoom no Fim (Frequência Alta - 10000 Hz)
plt.subplot(3, 1, 3)
plt.plot(t, sinal_chirp, color='green')
plt.xlim(T - 0.002, T)  # Mostra apenas os últimos 2 milissegundos
plt.title('Zoom no Fim: Frequência Alta (~10000 Hz)')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.grid(True)

plt.tight_layout()
plt.show()
```
